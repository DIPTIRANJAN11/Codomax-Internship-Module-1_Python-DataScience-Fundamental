```python
scores=[10,30,50,65,54]
for score in scores:
  if score>50:
    break
  print(score)

#skip missing data
data=[10,20,0,40,50,0]
for value in data:
  if value==0:
    continue
  print(value)
```

```python
def calculate(price):
  tax=price*0.10
  return tax
item_price=1000
my_tax=calculate(item_price)
print(f"tax ammount: ${my_tax}")
```

```python
def create_profile(name, role="Data Science Intern"):
    print(f"Name: {name} | Role: {role}")

# Uses the default role
create_profile("Alice")

# Overrides the default role
create_profile("Bob", "Senior Analyst")
```

```python
def calculate_average(score1, score2, score3):
    total = score1 + score2 + score3
    average = total / 3
    return average

student_avg = calculate_average(85, 90, 78)
print(f"Student Average: {student_avg:.2f}") # .2f formats to 2 decimal places
```

```python
#find the max value
def find_max(number):
highest=number[0]
for n in number:
  if n>highest:
    highest=n
return highest
scores=[10,20,30,40,65,82]
max_score=find_max(score)
print(f"highest score is:"{max_score})
```
