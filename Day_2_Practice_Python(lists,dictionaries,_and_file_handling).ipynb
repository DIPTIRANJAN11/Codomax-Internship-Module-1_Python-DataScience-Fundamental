```python
#Task: Read a file named data.txt that has one number per line. Calculate the sum of all the numbers.
# First, let's create a dummy file for this task
with open("numbers.txt", "w") as f:
    f.write("10\n20\n30\n40\n")

# Now, read and sum it
total = 0
with open("numbers.txt", "r") as file:
    for line in file:
        # We must convert the string to an integer before adding
        total = total + int(line)

print(f"The sum of numbers in the file is: {total}")
```

```python
#Task: Open a file named text.txt, read it, and count how many words are in the file. (Hint: Use .split() on the string).
# Create dummy file
with open("text.txt", "w") as f:
    f.write("Python is great for data science")

with open("text.txt", "r") as file:
    content = file.read()
    # .split() breaks the string into a list of words based on spaces
    words = content.split()
    word_count = len(words)

print(f"Total words in file: {word_count}")
```
